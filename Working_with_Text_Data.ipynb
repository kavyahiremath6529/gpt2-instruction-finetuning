```python
class SimpleTokenizerV1: #Creates a tokenizer, a tool that changes text into numbers and numbers back into text.
    def __init__(self, vocab): # Sets up the tokenizer with a vocabulary.
        #__init__ is the setup step that runs when you create a tokenizer.
        #Python runs __init__ automatically and gives the tokenizer its vocabulary dictionary
        
        self.str_to_int = vocab  # Saves the dictionary that maps text pieces to numbers.
        self.int_to_str = {i:s for s,i in vocab.items()}  # Makes a dictionary that maps numbers back to text.
    
    def encode(self, text): # Starts the function that changes text into numbers.
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text) # Splits the text at spaces and punctuation.
                                
        preprocessed = [
            item.strip() for item in preprocessed if item.strip() # Removes extra spaces and skips empty pieces.
        ]
        ids = [self.str_to_int[s] for s in preprocessed] # Looks up the number for each text piece.
        return ids
        
    def decode(self, ids): # Starts the function that changes numbers back into text.
        text = " ".join([self.int_to_str[i] for i in ids])  #Looks up each number and joins the text pieces with spaces.
        
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text) #Removes spaces that appear before punctuation.
        return text
    
    
    #Conclusion of the above code
#     We made a simple tokenizer:
# - It stores a dictionary that matches text pieces to numbers.
# - encode splits text into pieces and changes each piece into a number.
# - decode changes the numbers back into text and fixes spaces before punctuation.
```

```python
#We also need to adjust the tokenizer accordingly so that it knows when and how to use the new <unk> token
class SimpleTokenizerV2:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = { i:s for s,i in vocab.items()}
    
    def encode(self, text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        preprocessed = [
            item if item in self.str_to_int #if the string is not empty and present in the vocabulary, print item
            #else print "unknwon"
            else "<|unk|>" for item in preprocessed
        ]

        ids = [self.str_to_int[s] for s in preprocessed]
        return ids
        
    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        # Replace spaces before the specified punctuations
        text = re.sub(r'\s+([,.:;?!"()\'])', r'\1', text)
        return text
```

```python
# This tokenizer can replace words it doesn't know with the special <|unk|> token.
class SimpleTokenizerV2:  # Creates a new tokenizer type
    def __init__(self, vocab):  # Sets up the tokenizer using a word-to-number dictionary
        self.str_to_int = vocab  # Saves the dictionary that changes words into numbers
        self.int_to_str = {i: s for s, i in vocab.items()}  # Makes a dictionary that changes numbers back into words

    def encode(self, text):  # Changes text into a list of token numbers
        preprocessed = re.split(r'([,.:?_!"()\']|--|\s)', text)  # Splits text around punctuation, dashes, and spaces
        preprocessed = [item.strip() for item in preprocessed if item.strip()]  # Removes empty pieces and extra spaces
        preprocessed = [  # Starts a new list of cleaned-up tokens
            item if item in self.str_to_int  # Keeps the token if it exists in the vocabulary
            else "<|unk|>"  # Uses the unknown-token marker if the token isn't in the vocabulary
            for item in preprocessed  # Checks each token from the text
        ]  # Finishes making the token list
        ids = [self.str_to_int[s] for s in preprocessed]  # Changes each token into its number
        return ids  # Gives back the list of numbers

    def decode(self, ids):  # Changes a list of token numbers back into text
        text = " ".join([self.int_to_str[i] for i in ids])  # Looks up each token and joins them with spaces
        text = re.sub(r'\s+([,.:?!"()\'])', r'\1', text)  # Removes spaces before punctuation
        return text  # Gives back the readable text
```

```python
#Create dataset and dataloader that extract chunks from the input text dataset
# Import PyTorch tools for creating a dataset and loading it in batches.
from torch.utils.data import Dataset, DataLoader

# Create a dataset class for preparing text examples to train a GPT model.
class GPTDatasetV1(Dataset):
    # Set up the dataset using the text, tokenizer, chunk size, and step size.
    def __init__(self, txt, tokenizer, max_length, stride):
        self.input_ids = []  # Store the input token sequences here.
        self.target_ids = []  # Store the target sequences the model should predict.

        # Convert the whole text into token numbers.
        token_ids = tokenizer.encode(txt, allowed_special={"<|endoftext|>"})

        # Make sure there are enough tokens to create at least one example.
        assert len(token_ids) > max_length, "Number of tokenized inputs must be at least equal to max_length+1"

        # Move a window through the tokens to create examples, shifting by stride each time.
        for i in range(0, len(token_ids) - max_length, stride):
            
            #Stride is how many tokens the window moves forward each time it makes a new example.
            # Take max_length tokens to use as the model's input.
            #max_length is the window size; stride is how far the window moves each time.
            
            #Example: if max_length = 4 and stride = 2, each window contains 4 tokens, then moves forward 2 tokens.
            
            input_chunk = token_ids[i:i + max_length]

            # Take the following tokens as targets, shifted forward by one.
            target_chunk = token_ids[i + 1:i + max_length + 1]

            # Convert the input chunk into a PyTorch tensor and save it.
            self.input_ids.append(torch.tensor(input_chunk))

            # Convert the target chunk into a PyTorch tensor and save it.
            self.target_ids.append(torch.tensor(target_chunk))

    # Tell Python how many examples are in this dataset.
    def __len__(self):
        return len(self.input_ids)

    # Tell Python how to retrieve one example by its position.
    def __getitem__(self, idx):
        # Return the input tokens and their matching next-token targets.
        return self.input_ids[idx], self.target_ids[idx]
    
    # This class turns a long book into smaller examples. 
    # Each example has an input chunk and a target chunk shifted forward by one token, 
    # so the model can learn to predict what comes next.
```

```python
#A DataLoader is a helper that gives your model training examples in small groups called batches.

def create_dataloader_v1(  # Defines a function that prepares text examples for training.
    txt,  # The text to learn from.
    batch_size=4,  # How many examples to group together at a time.
    max_length=256,  # How many tokens each example contains.
    stride=128,  # How far the window moves to make the next example.
    shuffle=True,  # Mixes up the order of examples.
    drop_last=True,  # Drops the final batch if it has fewer examples than the others.
    num_workers=0  # Uses the main process to load data; 0 means no extra worker processes.
):
    # Set up the GPT-2 tokenizer, which turns text into token numbers.
    tokenizer = tiktoken.get_encoding("gpt2")

    # Turn the text into a dataset of input and target token chunks.
    dataset = GPTDatasetV1(txt, tokenizer, max_length, stride)

    # Create a tool that provides the dataset in batches.
    dataloader = DataLoader(
        dataset,  # The dataset to load examples from.
        batch_size=batch_size,  # The number of examples in each batch.
        shuffle=shuffle,  # Whether to mix up the examples before grouping them.
        drop_last=drop_last,  # Whether to discard a smaller-than-usual final batch.
        num_workers=num_workers  # How many worker processes help load the data.
    )

    return dataloader  # Give back the finished data loader.
```

```python
vocab_size = 50257
output_dim = 256

token_embedding_layer = torch.nn.Embedding(vocab_size, output_dim)

# - vocab_size = 50257 means there are 50,257 possible token IDs.
# - output_dim = 256 means each token ID will be represented by 256 numbers.
# - Embedding(...) creates a lookup table with 50,257 rows and 256 numbers in each row.
# For example, token ID 40 looks up row 40 and gets its 256-number representation.
```
